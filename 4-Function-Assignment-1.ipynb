```python
# Create a function that checks whether the given number is Odd or Even
```

```python
class OddEven:
    def OddEven():
        num=int(input("Enter a number: "))
        if((num%2)==1):
            print(num,"is odd number")
            output=f"{num} is odd number"
        else:
            print(num,"is Even number")
            output=f"{num} is Even number"
        
        return output
```

```python
OddEven.OddEven()
```

```python
# Create a function that tells elegibility of marriage for male and female according to their age limit like 21 for male and 18 for female
```

```python
class ElegiblityForMarriage:
    
    def Elegible():
        
        gen=input("Enter Your Gender: ")
        age=int(input("Enter Your Age: "))
        
        if(gen=="Male" and age>=21):
            print("Eligible")
            result="Eligible"
            
        elif(gen=="Femala" and age>=18):
            print("Eligible")
            result="Eligible"
            
        else:
            print("Not Eligible")
            result="Not Eligible"
            
        return result
```

```python
ElegiblityForMarriage.Elegible()
```

```python
eligibility_result = ElegiblityForMarriage.Elegible()
print("Result:", eligibility_result)
```

```python
#print area and perimeter of triangle using class and functions
```

```python
class triangle:
    
    def triangle():
        height=float(input("Enter the Height: "))
        breadth=float(input("Enter the Breadth: "))
        height1=float(input("Enter the Height1: "))
        height2=float(input("Enter the Height2: "))
        
        area=(height*breadth)/2
        perimeter=height1+height2+breadth
        
        print("Area of the Triangle: ",area)
        print("Perimeter of the Triangle: " ,perimeter)
        
        return area,perimeter
```

```python
triangle.triangle()
```
